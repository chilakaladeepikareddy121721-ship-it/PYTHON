# List
* String multiple values
* we will use ([])
* it is mutable
* orderd

```python
a =  ["deepika","123",23]
print(a)
```

```python
type(a)
```

```python
a = ["deepika",123,23]
a[0]="abc"
print(a)
```

```python
a = ["deepika",123,23]
a.append( "abc")
print(a)
```

```python
a = ["deepika",123,23]
a.remove(123)
print(a)
```

```python
a = ["deepika",123,23]
a.extend([1,2,3])
print(a)
```

# Dictionary

```python
a = {
    "Name":"Deepika",
    "age" :20,
    "Gender" :"Female"
}
print(a)
```

```python
type(a)
```
